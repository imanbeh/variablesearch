```python
catalog = pd.read_csv("simbad_catalog_47_otype.txt",delimiter ="|",skipinitialspace=True,
                      names= ["Main_ID","OTYPE","RA","DEC","DIST"]) # read in simbad catalog
```

```python
catalog.head(5)
```

```python
#drop first few rows
catalog = catalog.drop([0,1,2],axis=0)
```

```python
catalog.columns
```

```python
#reset index to start at zero
catalog = catalog.reset_index()
catalog = catalog.drop('index',axis=1)
```

```python
# convert RA, DEC and DIST to floats
catalog['RA'] = catalog['RA'].astype(float)
catalog['DEC'] = catalog['DEC'].astype(float)
catalog['DIST'] = catalog['DIST'].astype(float)
```

```python
catalog.head(5)
```

```python
# open image data
fits_475 = fits.open('MAST_CMD/J6LL01010_475/j6ll01010_drz.fits')
fits_550 = fits.open('MAST_CMD/J6LL01YGQ_555/j6ll01ygq_drz.fits')
```

```python
# get wcs data of each
w_475 = WCS(fits_475[1].header)
w_550 = WCS(fits_550[1].header)
```

```python
print(catalog['RA'][0],catalog['DEC'][0])
```

```python
coord = SkyCoord(catalog['RA'][0],catalog['DEC'][0], unit = "deg")
coord
```

```python
a = w_475.world_to_pixel(coord)
print(a[0])
```

```python
catalog.shape[0]
```

```python
#make empty ra and dec to fill
x_475 = np.zeros_like(catalog['RA'])
y_475 = np.zeros_like(catalog['DEC'])

for i in range(catalog.shape[0]):
    x_475[i],y_475[i] = w_475.world_to_pixel(SkyCoord(catalog['RA'][i],catalog['DEC'][i], unit = "deg"))

catalog['X_475'] = x_475
catalog['Y_475'] = y_475
```

```python
catalog.head(10)
```

```python
#make empty ra and dec to fill
x_550 = np.zeros_like(catalog['RA'])
y_550 = np.zeros_like(catalog['DEC'])

for i in range(catalog.shape[0]):
    x_550[i],y_550[i] = w_550.world_to_pixel(SkyCoord(catalog['RA'][i],catalog['DEC'][i], unit = "deg"))

catalog['X_550'] = x_550
catalog['Y_550'] = y_550
```

```python
catalog.head(15)
```

```python
# only keep stars
catalog = catalog[catalog['OTYPE'].str.contains('\*')]
```
