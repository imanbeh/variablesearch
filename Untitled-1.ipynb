```python
def find_err(min, mag, max):
    '''
    find max and min distance from mean
    '''
    #print(np.abs(mag.values[i]-min[i]) == np.abs(mag.values[i]-max[i]))

    ds1s2 = pd.DataFrame(data = {"Min": min, "Mag": mag, "Max": max})
    print("Mag-Min")
    print(np.abs(ds1s2["Mag"]-ds1s2["Min"]))
    print("Max-Mag")
    print(np.abs(ds1s2["Mag"]-ds1s2["Max"]))

    for i in range(3):
        if (np.abs(mag.values[i]-min[i]) == np.abs(mag.values[i]-max[i])):
            print("Errors Match")
        else:
            print(np.abs(mag.values[i]-min[i]) - np.abs(mag.values[i]-max[i]))
        print(mag.values[i], "+-", np.abs(mag.values[i]-min[i]))
        print(mag.values[i], "+-", np.abs(max[i]-mag.values[i]))
    
```
